**insert**

```python
- difference between append and insert:
- insert means add an element to the list in between other elements
- append means add an element to the list at last
```

```python
l1=[10,20,30,40,50,'Apple','Banana','Cherry',40,50]
# i want to insert 100 after apple
i1=l1.index('Apple')
l1.insert(i1+1,100)
l1
```

```python
l1=[10,20,30,40,50,'Apple','Banana','Cherry',40,50]
l1.append(100)
l1
```

**pop vs remove**

```python
- pop will remove last element from list
- if we mention index in method bracket, then pop will delete that index element from the list
- remove will delete particular element from the list,which was mentioned in method brackets
- but remove will delete first occurence of that element
```

```python
l1=[10,20,30,40,50,'Apple','Banana','Cherry',40,50]
l1.pop()
l1
```

```python
l1=[10,20,30,40,50,'Apple','Banana','Cherry',40,50]
l1.remove('Apple')
l1
```

```python
l1=[10,20,30,40,50,'Apple','Banana','Cherry',40,50]
l1.remove(40)
l1
```

**in place**
- when we perform the operations sometimes the values will not overwrite
- if we see **inplace** word or **inplace=True**
- this indicates the output will be overwrite in corresponding variable

```python
l1=[10,20,30,40,50]
l1.reverse()
l1
```

```python
l1=[10,20,30,40,50]
reversed(l1)
print(l1)
```

```python
l1=[10,20,30,40,50]
for i in reversed(l1):
    print(i)
```
