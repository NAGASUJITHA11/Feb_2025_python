## List comprehension
- List comprehension is a way to write list operations in a single line
- when we apply for loop and if-else conditional statements then the code compilation becomes slow
- which means it takes much time
- inorder to avoid the time complexity we will use list comprehension
- list comprehension is a single line code

**case-1: only for loop**

```python
list1=[]
for i in range(10):
    list1.append(i)
list1
```

**syntax**

```python
list1=[<code should write here>]
# inside code no need of append
list1=[<output var> <for loop>]
```

**case-2: for loop if condition**

```python
# only even number
even_list=[]
for i in range(1,10):
    if i%2==0:
        even_list.append(f'even:{i}')
even_list
```

**syntax**

```python
[<output> <for loop> <if condition>]
```

```python
[f'even:{i}' for i in range(1,10) if i%2==0]
```

**case-3: for-if-else**

```python
case-1: for loop: [<output> <for loop>]
case-2: for-if : [<output> <for loop> <if condition>]
case-3: for-if-else : [<if output> <if condition> else <else output> <for loop>]
```

```python
num1=eval(input('enter a number:'))
if num1%2==0:
    print(f'even:{num1}')
else:
    print(f'odd:{num1}')
```

```python
l=[]
for i in range(1,10):
    if i%2==0:
        l.append(f'even:{i}')
    else:
        l.append(f'odd:{i}')
l
```

```python
[f'even:{i}' if i%2==0 else f'odd:{i}' for i in range(1,10) ]
```

**case-4: for-if-elif-else**

```python
- in list comprehension we dont have **elif** representation
- elif also represent as **else+if**
- in elif we have **elif condition** and **elif output**
- else elif_output elif_condition
```

```python
case-1: for loop: [<output> <for loop>]
case-2: for-if : [<output> <for loop> <if condition>]
case-3: for-if-else : [<if output> <if condition> else <else output> <for loop>]
case-4: for-if-elif-else:

[<if out> <if cond> else <elif out> <elif cond> else <else out> <for loop>]

**Note**: instead of elif term write if
```

```python
# >85 A grade   60-85 B grade    <60 C grade
l=[45,55,65,75,85,95]
for i in l:
    if i>85:
        print(f'{i}:A grade')
    elif i>60:
        print(f'{i}: B grade')
    else:
        print(f'{i}: C grade')
```

```python
#[<if out> <if cond> else <elif out> <elif cond> else <else out> <for loop>]
l=[45,55,65,75,85,95]
[f'{i}:A grade' if i>85 else f'{i}:B grade' if i>60 else f'{i}: C grade' for i in l]
```
